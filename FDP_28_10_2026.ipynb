```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load dataset
data = pd.read_csv('//content/Faculty_Experience_Salary_Linear_Regression.csv')

# Select X and Y
X = data[['Experience_Years']]
y = data['Salary_Lakhs']

# Create and train model
model = LinearRegression()
model.fit(X, y)

# Predict salary
y_pred = model.predict(X)

# Print equation values
print("Intercept:", model.intercept_)
print("Slope:", model.coef_[0])

# Predict salary for 5 years experience
print("Salary for 5 years experience:",
      model.predict([[5]])[0], "Lakhs")

# Plot
plt.scatter(X, y)
plt.plot(X, y_pred)
plt.xlabel("Experience (Years)")
plt.ylabel("Salary (Lakhs)")
plt.title("Experience vs Salary")
plt.show()
```
