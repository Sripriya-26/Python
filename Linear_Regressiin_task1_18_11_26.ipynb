```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x=[[1],[2],[3]]
y=[[5],[10],[15]]
model=LinearRegression()
model.fit(x,y)
print(model.predict([[10]]))
m=model.coef_[0]
c=model.intercept_
print("slope:", m)
print("y intercept: ",c)
```
