1.Create a dataset students,marks

```python
import pandas as pd

df= pd.DataFrame({

                  "Student":["A","B","C","D","E","F","G","H","i","m"],
                  "Marks":[45,48,50,52,49,51,47,95,89,33]
})
print(df)
```

Descriptive Statistics

```python
print(df['Marks'].mean())
df['Marks'].median()
df['Marks'].mode()
df['Marks'].std()
df['Marks'].min()
df['Marks'].max()
```

find Q1,Q3,IQR

```python
Q1=df['Marks'].quantile(0.25)
Q3=df['Marks'].quantile(0.75)
IQR=Q3-Q1
print(Q1)
print(Q3)
print(IQR)
```

```python
lower=Q1-1.5*IQR
upper=Q3+1.5*IQR
print(lower)
print(upper)
```

```python
outliers=[]
for i in df['Marks']:
    if i<lower or i>upper:
        outliers.append(i)
print(outliers)
```

```python
import matplotlib.pyplot as plt
plt.boxplot(df['Marks'])
plt.show()
```
